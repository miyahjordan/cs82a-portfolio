```python
import pandas as pd
df = pd.read_csv('netflix_titles.csv')
df.shape
```

```python
df.head()
```

* Looking at missing values



```python
df.isna().sum()
```


* Director, cast, country, date_added, rating, and duration are missing
* To see how movies and TV shows compare, it would be beneficial to replace country data with the most common entry
* Dropping all of the null values may result in removing valuable data, so I will replace the nulls in the director and cast category with another text value

```python
common_country = df['country'].mode()[0]
df['country'] = df['country'].fillna(common_country)
print(common_country)
```

```python
df['director'] = df['director'].fillna('None Stated')
df['cast'] = df['cast'].fillna('None Stated')
df.head(10)
```

```python
df.dropna()
```

```python
df.isna().sum()
```

* Checking for duplicates

```python
print(df.duplicated().sum())
```

* Checking data types to make sure they align

```python
df.dtypes
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format = 'mixed')
df['month_num_added'] = df['date_added'].dt.month
df['month_name_added'] = df['date_added'].dt.month_name()
df['year_added'] = df['date_added'].dt.year
df.head()
```

```python
df.dtypes
```

* Calculating ratio of Movies and TV shows

```python
df.groupby('type').agg(movie_tv_ratios = ('type', 'count'))
```

```python
ratio_percentages = round((df.groupby('type').agg(movie_tv_ratios = ('type', 'count'))/8807) * 100, 2)
print(ratio_percentages)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

size_of_groups = [69.62, 30.38]
plt.pie(size_of_groups, labels = ['Movies \n 69.62%', 'TV Shows \n 30.38%'], textprops = {'family':'serif'}, colors = ['#E50914', '#221f1f'])
plt.suptitle('Ratio of Movies and TV Shows', fontfamily = 'serif', fontsize = 20)
plt.title('A majority of the offerings on Netflix are Movies', fontfamily = 'serif', fontsize = 12)
center_circle = plt.Circle((0,0), 0.7, color = 'white')
p = plt.gcf()
p.gca().add_artist(center_circle)

plt.show()
```

* How movie length changed over time

```python
movies_df = df[df['type'] == 'Movie'].copy()
movies_df['duration'] = movies_df['duration'].str.replace(' min', '')
movies_df['duration'] = pd.to_numeric(movies_df['duration'])
movies_df.head()
```

```python
movie_duration_over_time = movies_df.groupby('year_added').agg(movie_length = ('duration', 'mean'))
print(movie_duration_over_time)
```

```python
plt.rcParams['font.family'] = 'serif'
sns.lineplot(data = movie_duration_over_time, x = 'year_added', y = 'movie_length', color = '#E50914')
plt.suptitle('Movie Length Over Time', fontsize = 16)
plt.title('The average duration of movies have gotten higher since 2012', fontsize = 10)
plt.xlabel('Release Year')
plt.ylabel('Average Movie Duration (minutes)')
plt.show()
```

* How TV Show length changed over time

```python
tv_show_df = df[df['type'] == 'TV Show'].copy()
tv_show_df['duration'] = tv_show_df['duration'].str.replace(' Seasons', '').str.replace(' Season', '')
tv_show_df['duration'] = pd.to_numeric(tv_show_df['duration'])
tv_show_df.head()
```

```python
tv_show_duration_over_time = tv_show_df.groupby('year_added').agg(tv_show_duration = ('duration', 'mean'))
print(tv_show_duration_over_time)
```

```python
plt.rcParams['font.family'] = 'serif'
sns.lineplot(data = tv_show_duration_over_time, x = 'year_added', y = 'tv_show_duration', color = '#E50914')
plt.suptitle('TV Show Length Over Time', fontsize = 16)
plt.title('The average duration of TV shows have not recovered since 2014 peak', fontsize = 10)
plt.xlabel('Release Year')
plt.ylabel('Average TV Show Duration (seasons)')
plt.show()
```

* content across countries

```python
movies_by_year = movies_df.groupby('year_added').agg(movies_added = ('type', 'count'))
tv_shows_by_year = tv_show_df.groupby('year_added').agg(tv_shows_added = ('type', 'count'))
print(movies_by_year, tv_shows_by_year)
```

```python
plt.rcParams['font.family'] = 'serif'
sns.lineplot(data = movies_by_year, x = 'year_added', y = 'movies_added', color = '#E50914')
sns.lineplot(data = tv_shows_by_year, x = 'year_added', y = 'tv_shows_added', color = '#221f1f')
ax = plt.gca()
fig = plt.gcf()
ax.yaxis.tick_right()
plt.suptitle('Movies & TV Shows Added Over Time', size = 16)
plt.title('More movies are added to Netflix compared to TV shows')
plt.xlabel('Year')
plt.ylabel('')
fig.text(0.2, 0.2, 'Movie', fontweight = 'bold', fontsize = 12, color = '#E50914')
fig.text(0.3, 0.2, '| TV Show', fontweight = 'bold', fontsize = 12, color = '#221f1f')
plt.show()
```
