```python
temperatures = [10, 16, 23, 28, 14, 22, 30]
for temperature in temperatures:
  if temperature < 15:
    print(f"{temperature} degrees Celsius: Cold")
  elif temperature >= 15 and temperature <=25:
    print(f"{temperature} degrees Celsius: Warm")
  else:
    print(f"{temperature} degrees Celsius: Hot")
```
