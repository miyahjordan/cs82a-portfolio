For the homework, we're only concerned with the length column. Looking at that column, the mean is around 1567.47 while the median is 1300.0. Judging by the distribution of the lengths in the data, I'm more inclinded to used the median to replace the missing values rather than the mean. While 1567.47 isn't close to the maximum value, so far, there is only 2 values in the data over 1300 which leads me to believe theyre skewing the average a little higher. Replacing the missing values will probably give me a very unrealistic outlook, while using the median will skew the data closer to the already established data points.

```python
fill_value = df['length'].median()
df['length'] = df['length'].fillna(fill_value)
```
