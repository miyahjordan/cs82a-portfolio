```python
import pandas as pd
import statistics
df = pd.read_csv('sales_clean.csv')
df.shape
```

```python
print('mean price:', round(df["price"].mean(), 2))
print('median price:', round(df["price"].median(), 2))
print('price standard deviation:', round(df["price"].std(), 2))
print('mean quantity:', round(df["qty"].mean(), 2))
print('median quantity:', round(df["qty"].median(), 2))
print('quantity standard deviation:', round(df["qty"].std(), 2))
```

I would report the median rather than the mean because is a very large gap betwen the two which lets me know there are some outling values that are heavily affecting the mean. If I were to replace the price with the mean, it would probably heavily inflate the prices.
