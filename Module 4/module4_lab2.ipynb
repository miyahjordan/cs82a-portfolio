```python
import pandas as pd
df = pd.read_csv('allbacks.csv')
df.shape
```

```python
df.head(15)
```

```python
df.sort_values(by=['weight'])
print('median weight:', df['weight'].median(), 'median cover:', 'pb')
```

The middle value of the data after being sorted (by weight) is 725.0 which belongs to a book with a paperback cover.

```python
print('mean weight:', round(df['weight'].mean(), 2))
```

The average weight of all the books is 706.67 grams

```python
print('weights_by_cover:', round(df.groupby('cover')['weight'].mean(), 2))
```

The average weight of hardcover books are 796.43 grams and the average weight of paperback books are 638.12 grams.
