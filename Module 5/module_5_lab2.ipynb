```python
import pandas as pd
df = pd.read_csv('custdata.tsv', sep='\t')
df.shape
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df.dropna(subset = ['housing.type'], inplace = True)
df.isna().sum()
```

```python
df.shape
```

I excluded 56 rows.

```python
df['custid'] = df['custid'].astype('int64')
df['customer count'] = df.count['custid']
```

```python
df.groupby('housing.type')['custid'].mean().sort_values()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.barplot(data = df, x = 'housing.type', y = 'custid')
plt.title('Housing Type by Customer')
plt.xlabel('Housing Type')
plt.ylabel('Customers')
plt.show()
```
