```python
import pandas as pd
df = pd.read_csv('ratings.tsv', sep = '\t')
df.shape
```

```python
df.head()
```

```python
import sklearn
from sklearn.model_selection import train_test_split
X = df[["food", "ambience", "rating"]]
y = df["restaurant"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
print('X_train:', X_train, 'X_test:', X_test, 'y_train:', y_train, 'y_test:', y_test)
```

```python
len(X_train)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

```python
print(model.coef_, model.intercept_)
```

Ranked Coefficients:
1. food = 0.92
2. ambience = -0.07
3. rating = -0.96


With ratings being the lowest coeffeficient, I would say the a struggling restaurant should focus on bringing ratings up by fixing the food. Food seems to have a very high correlation, so by improviung the food, you can improve your ratings.
