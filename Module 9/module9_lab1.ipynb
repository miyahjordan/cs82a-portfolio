```python
def pounds_to_kilograms(pounds):
  kilograms = round((pounds * 2.205), 2)
  print(kilograms, 'kilograms')
```

Converting my weight to kilograms:

```python
pounds_to_kilograms(119.5)
```

Converting average weight of a newborn baby:

```python
pounds_to_kilograms(7.3)
```

Converting average weight of a semi-truck:

```python
pounds_to_kilograms(20000)
```
