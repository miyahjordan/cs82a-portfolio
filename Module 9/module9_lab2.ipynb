How to migrate from Jupyter Notebook to Google Colab:
1. Set up a google account to access Google Colab
2. Notebooks will transfer over, but if they are connected files you will need to upload them to the Colab session, and python 3 is automatically integrated
3. *To upload notebooks*: Select upload notebook from the Colab homepage
4. *To create a new notebook*: Select new notebook at the Colab homepage
5. *To upload files to Colab notebook*: Select files from sidebar and select upload to session storage. Files will be removed from the session when it closes.
6. Colab files can be saved as .py or .ipynb
7. Risk: All Colab files are tied to your Google account. In the case that you get hacked or the account is removed, the Colab files will not be recoverable.
