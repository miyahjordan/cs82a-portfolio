#Cleaning Log

1. checking for missing data

    There were 12 missing rows from the price column. Price is a needed value for sales data, so I replaced the missing data within the column with the median value of the prices. I used the median because using the mean may overinflate price
    
    The 12 prices were filled with 56.28500428082192

2.  check for duplicates
    
    Originally there were 300 rows loaded into the data, after checking for duplicates there were 8 duplicate rows. Duplicates overinflate the data and cause problems with calculations and analysis, so I removed them.
    
    300 rows loaded; 8 duplicates removed; there are now 292 rows in the data

3. changing data types

    The date and zip columns were under the worng data types, text for date and integer for zip. Also, there were different date formats within the that column. Zip codes cannot be listed as a number in the case they start with a 0 AND they must be 5 charcters long. Dates need to be standardized and formatted correctly for prosperity and to make analysis easier.
    
    Changed zip column from integer to string; made zip codes 5 characters long; changed date column from text to date; standardized dates

4. dealing with negative values

    There are some negative values in the quantity column, but we don't know what they represent. Since the data tells us the items sold from a specific store, I interpreted the negative data as item returns.

    negatives kept with the data; stored under a variable named returns

5. Cleaning data is important because any misstep such as keeping duplicates or removing the wrong columns could cause issues wsuch as having too much data, too little, or having answers that don't make sense.
